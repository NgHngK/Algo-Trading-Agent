```python
from PIL import Image
import numpy as np

# 1. Load your image (update the path to wherever your image is stored)
img = Image.open('dog.jpeg')
arr = np.array(img)

# 2. Print the RGB triplet for a given pixel, e.g. top‑left:
y, x = 0, 0
r, g, b = arr[y, x]
print(f"Pixel ({y},{x}): R={r}, G={g}, B={b}")

# 3. (Optional) Loop over a small patch, say 5×5 in the top‑left corner:
for y in range(5):
    for x in range(5):
        r, g, b = arr[y, x]
        print(f"Pixel ({y},{x}): R={r}, G={g}, B={b}")
```
