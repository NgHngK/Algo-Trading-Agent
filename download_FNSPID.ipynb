```python
from pathlib import Path
from huggingface_hub import hf_hub_download

REPO_ID = "Zihan1004/FNSPID"
DATA_DIR = Path("data/FNSPID")

DATA_DIR.mkdir(parents=True, exist_ok=True)

# Download financial news

news_path = hf_hub_download(
    repo_id=REPO_ID,
    filename="Stock_news/nasdaq_exteral_data.csv",
    repo_type="dataset",
    local_dir=DATA_DIR,
)

print("News downloaded to:")
print(news_path)

# Download historical stock prices

price_path = hf_hub_download(
    repo_id=REPO_ID,
    filename="Stock_price/full_history.zip",
    repo_type="dataset",
    local_dir=DATA_DIR,
)

print("\nStock prices downloaded to:")
print(price_path)
```

```python
from pathlib import Path
import zipfile

zip_path = Path("data/FNSPID/Stock_price/full_history.zip")
extract_dir = Path("data/FNSPID/Stock_price/full_history")

extract_dir.mkdir(parents=True, exist_ok=True)

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_dir)

print("Extracted to:", extract_dir)
```
